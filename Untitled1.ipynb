```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession.builder.appName('test').getOrCreate()
```

```python
spark.sparkContext.setLogLevel("Error")
```

```python
## reading the fraudTrain.csv
file_path = ('./raw_data/fraudTrain.csv')
bronze_df = spark.read.csv(file_path, header=True, inferSchema=True)
bronze_df.select(col('is_fraud')).distinct().count().show()
```

```python
bronze_df.groupBy(col('is_fraud')).count().show()
```
